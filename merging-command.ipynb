```python
import nbformat

files = [
    "Exp1_DDA.ipynb",
    "Exp2_BresenhamsLDA.ipynb",
    "Exp3_MidPointCircle.ipynb",
    "Exp4_BresenhamsCircle.ipynb",
    "Exp5_EllipseDA.ipynb",
    "Exp6_2DTransformations.ipynb",
    "Exp7_CohenSutherland.ipynb",
    "Exp8_LiangBarsky.ipynb",
    "Exp9_BezierCurve.ipynb",
    "Exp10_B-SplineCurve.ipynb",
    "Exp11_BezierBlending.ipynb",
    "Exp12_TextFonts.ipynb"
]
merged = nbformat.v4.new_notebook()

for file in files:
    with open(file, "r", encoding="utf-8") as f:
        notebook = nbformat.read(f, as_version=4)
        merged.cells.extend(notebook.cells)

with open("All_Graphics_Lab.ipynb", "w", encoding="utf-8") as f:
    nbformat.write(merged, f)

print("Notebooks merged successfully!!")
```
